```python
# Load the supply chain dataset

file_path = "../data/raw/dynamic_supply_chain_logistics_dataset_with_country.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

```python
# Check number of rows and columns

print("Dataset Shape:", df.shape)
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
# Display first 5 records

df.head()
```

```python
df.shape
```

```python
# Display all column names

print("Total Columns:", len(df.columns))

for i, column in enumerate(df.columns, start=1):
    print(i, column)
```

```python
# Display dataset information and data types

df.info()
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
# Check duplicate records

duplicates = df.duplicated().sum()

print("Total Duplicate Rows:", duplicates)
```

```python
# Statistical summary of numerical columns

df.describe().T
```

```python
# Check unique values for categorical columns

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print("\nColumn:", column)
    print("Unique Values:", df[column].nunique())
    print(df[column].value_counts().head(10))
```

```python
# Check data types of all columns

print("DATA TYPES")
print("=" * 50)

print(df.dtypes)
```

```python
# Check class distribution

print("Risk Classification Distribution:")
print(df["risk_classification"].value_counts())

print("\nPercentage Distribution:")
print(
    df["risk_classification"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# Check number of unique products, suppliers and countries

print("Unique Products:",
      df["product_id"].nunique())

print("Unique Suppliers:",
      df["supplier_id"].nunique())

print("Unique Supplier Countries:",
      df["supplier_country"].nunique())


      
```
