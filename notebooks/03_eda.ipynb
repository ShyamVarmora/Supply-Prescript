```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
file_path = "../data/processed/supply_chain_cleaned.csv"

df = pd.read_csv(file_path)

print("Cleaned dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nShape:", df.shape)

df.head()
```

```python
risk_counts = df["risk_classification"].value_counts()

print(risk_counts)

plt.figure(figsize=(7,5))
risk_counts.plot(kind="bar")
plt.title("Risk Classification Distribution")
plt.xlabel("Risk Class")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["delay_probability"], bins=30)

plt.title("Distribution of Delay Probability")
plt.xlabel("Delay Probability")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["delivery_time_deviation"], bins=30)

plt.title("Delivery Time Deviation Distribution")
plt.xlabel("Delivery Time Deviation")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["supplier_reliability_score"],
    df["delay_probability"],
    alpha=0.2
)

plt.title("Supplier Reliability vs Delay Probability")
plt.xlabel("Supplier Reliability Score")
plt.ylabel("Delay Probability")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["lead_time_days"],
    df["delay_probability"],
    alpha=0.2
)

plt.title("Lead Time vs Delay Probability")
plt.xlabel("Lead Time Days")
plt.ylabel("Delay Probability")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["route_risk_level"],
    df["delay_probability"],
    alpha=0.2
)

plt.title("Route Risk Level vs Delay Probability")
plt.xlabel("Route Risk Level")
plt.ylabel("Delay Probability")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["weather_condition_severity"],
    df["delay_probability"],
    alpha=0.2
)

plt.title("Weather Severity vs Delay Probability")
plt.xlabel("Weather Condition Severity")
plt.ylabel("Delay Probability")

plt.show()
```

```python
print("=" * 60)
print("ALL COLUMNS IN CLEANED DATASET")
print("=" * 60)

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
print("=" * 70)
print("NUMERICAL STATISTICAL SUMMARY")
print("=" * 70)

numeric_cols = df.select_dtypes(include=np.number).columns

print("Total Numerical Columns:", len(numeric_cols))

display(df[numeric_cols].describe().T)
```

```python
import seaborn as sns

corr = df[numeric_cols].corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    annot=True,
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Supply Chain Variables")
plt.tight_layout()
plt.show()
```

```python
risk_delay = (
    df.groupby("risk_classification")["delay_probability"]
      .mean()
      .sort_values(ascending=False)
)

print(risk_delay)

plt.figure(figsize=(8, 5))
risk_delay.plot(kind="bar")

plt.title("Average Delay Probability by Risk Classification")
plt.xlabel("Risk Classification")
plt.ylabel("Average Delay Probability")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
risk_deviation = (
    df.groupby("risk_classification")["delivery_time_deviation"]
      .mean()
      .sort_values(ascending=False)
)

print(risk_deviation)

plt.figure(figsize=(8, 5))
risk_deviation.plot(kind="bar")

plt.title("Average Delivery Time Deviation by Risk Class")
plt.xlabel("Risk Classification")
plt.ylabel("Average Delivery Time Deviation")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
high_risk = df[df["risk_classification"] == "High Risk"]

country_risk = (
    high_risk["supplier_country"]
    .value_counts()
    .head(10)
)

print(country_risk)

plt.figure(figsize=(10, 6))
country_risk.sort_values().plot(kind="barh")

plt.title("Top 10 Supplier Countries with High-Risk Records")
plt.xlabel("Number of High-Risk Records")
plt.ylabel("Supplier Country")
plt.tight_layout()
plt.show()
```

```python
supplier_risk = (
    high_risk["supplier_id"]
    .value_counts()
    .head(10)
)

print(supplier_risk)

plt.figure(figsize=(10, 6))
supplier_risk.sort_values().plot(kind="barh")

plt.title("Top 10 Suppliers with High-Risk Records")
plt.xlabel("Number of High-Risk Records")
plt.ylabel("Supplier ID")
plt.tight_layout()
plt.show()
```

```python
product_risk = (
    high_risk["product_id"]
    .value_counts()
    .head(10)
)

print(product_risk)

plt.figure(figsize=(10, 6))
product_risk.sort_values().plot(kind="barh")

plt.title("Top 10 Products with High-Risk Records")
plt.xlabel("Number of High-Risk Records")
plt.ylabel("Product ID")
plt.tight_layout()
plt.show()
```

```python
reliability_risk = (
    df.groupby("risk_classification")["supplier_reliability_score"]
      .mean()
      .sort_values()
)

print(reliability_risk)

plt.figure(figsize=(8, 5))
reliability_risk.plot(kind="bar")

plt.title("Average Supplier Reliability by Risk Class")
plt.xlabel("Risk Classification")
plt.ylabel("Average Supplier Reliability Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
route_risk = (
    df.groupby("risk_classification")["route_risk_level"]
      .mean()
      .sort_values(ascending=False)
)

print(route_risk)

plt.figure(figsize=(8, 5))
route_risk.plot(kind="bar")

plt.title("Average Route Risk Level by Risk Classification")
plt.xlabel("Risk Classification")
plt.ylabel("Average Route Risk Level")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
weather_risk = (
    df.groupby("risk_classification")["weather_condition_severity"]
      .mean()
      .sort_values(ascending=False)
)

print(weather_risk)

plt.figure(figsize=(8, 5))
weather_risk.plot(kind="bar")

plt.title("Average Weather Severity by Risk Classification")
plt.xlabel("Risk Classification")
plt.ylabel("Average Weather Severity")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
disruption_risk = (
    df.groupby("risk_classification")["disruption_likelihood_score"]
      .mean()
      .sort_values(ascending=False)
)

print(disruption_risk)

plt.figure(figsize=(8, 5))
disruption_risk.plot(kind="bar")

plt.title("Average Disruption Likelihood by Risk Classification")
plt.xlabel("Risk Classification")
plt.ylabel("Average Disruption Likelihood Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
lead_risk = (
    df.groupby("risk_classification")["lead_time_days"]
      .mean()
      .sort_values(ascending=False)
)

print(lead_risk)

plt.figure(figsize=(8, 5))
lead_risk.plot(kind="bar")

plt.title("Average Lead Time by Risk Classification")
plt.xlabel("Risk Classification")
plt.ylabel("Average Lead Time (Days)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
shipping_risk = (
    df.groupby("risk_classification")["shipping_costs"]
      .mean()
      .sort_values(ascending=False)
)

print(shipping_risk)

plt.figure(figsize=(8, 5))
shipping_risk.plot(kind="bar")

plt.title("Average Shipping Cost by Risk Classification")
plt.xlabel("Risk Classification")
plt.ylabel("Average Shipping Cost")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# EDA Conclusion

The exploratory data analysis identified important patterns related to supply chain risk and delivery performance.

Key observations:

1. The dataset contains a considerably higher number of High Risk records compared with Moderate Risk and Low Risk records.

2. Delay probability is an important indicator of supply chain risk and will be considered during predictive modeling.

3. Delivery time deviation provides information about how actual delivery performance differs from expected delivery performance.

4. Supplier reliability, route risk, weather severity, lead time, and disruption likelihood are important operational factors for analyzing supply chain disruptions.

5. Supplier-level analysis helps identify suppliers associated with a high number of risky transactions.

6. Country-level analysis helps identify geographical concentrations of supply chain risk.

7. Product-level analysis identifies products that frequently appear in high-risk supply chain situations.

8. Correlation analysis helps determine which numerical variables have stronger relationships with delay and supply chain risk.

9. The EDA results will be used for feature selection and predictive model development.

10. The next phase will develop a machine-learning model for predicting supply chain delay/risk and later connect those predictions with prescriptive optimization.
```
